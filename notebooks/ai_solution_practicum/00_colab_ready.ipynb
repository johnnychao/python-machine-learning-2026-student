## 1. 先確認執行環境

GPU 是加速選項，不是通關條件。CNN、RNN 與風格轉換若用 CPU，
需要多等一點；RL 則只使用 Python。


```python
import importlib
import json
import platform
import sys
from datetime import datetime, timezone
from pathlib import Path

required_packages = {
    "numpy": "numpy",
    "pandas": "pandas",
    "scikit-learn": "sklearn",
    "matplotlib": "matplotlib",
    "Pillow": "PIL",
    "TensorFlow": "tensorflow",
}

package_status = {}
for display_name, import_name in required_packages.items():
    module = importlib.import_module(import_name)
    package_status[display_name] = getattr(module, "__version__", "available")

import tensorflow as tf

gpu_devices = tf.config.list_physical_devices("GPU")
compute = gpu_devices[0].name if gpu_devices else "CPU（可繼續）"

print("Python:", sys.version.split()[0])
print("系統:", platform.platform())
print("運算裝置:", compute)
print("套件:", package_status)
```

## 2. 產生準備紀錄

看到 READY 就可以進入四個案例。若套件匯入失敗，先在 Colab
選單執行「執行階段 → 重新啟動工作階段」，再重跑一次。


```python
readiness = {
    "status": "READY",
    "checked_at_utc": datetime.now(timezone.utc).isoformat(timespec="seconds"),
    "python": sys.version.split()[0],
    "compute": compute,
    "packages": package_status,
    "data_access": "Notebook 內公開直連",
    "student_flow": "四案快速體驗後，選一案完整實作",
}

report_path = Path("/content/colab_readiness.json")
report_path.write_text(
    json.dumps(readiness, ensure_ascii=False, indent=2),
    encoding="utf-8",
)
print("READY：可以開始綜合實作")
print("紀錄檔:", report_path)
```
