## 2. 檢查 runtime 與主要套件


```python
import importlib
import os
import platform
import sys
from datetime import datetime, timezone

package_names = ["numpy", "pandas", "sklearn", "tensorflow", "torch", "kagglehub", "gradio"]
versions = {}
for package_name in package_names:
    module = importlib.import_module(package_name)
    versions[package_name] = getattr(module, "__version__", "unknown")

gpu_name = "CPU only"
try:
    import tensorflow as tf
    devices = tf.config.list_physical_devices("GPU")
    if devices:
        gpu_name = devices[0].name
except Exception as exc:
    gpu_name = f"TensorFlow GPU check failed: {type(exc).__name__}"

print("Python:", sys.version.split()[0])
print("Platform:", platform.platform())
print("Compute:", gpu_name)
print("UTC checked at:", datetime.now(timezone.utc).isoformat(timespec="seconds"))
print("Packages:", versions)
```

## 3. 安全讀取 Colab Secret（選用）

公開 Dataset 模式不一定需要 Token。只有當 Kaggle 回覆需要認證或同意資源條款時才設定。


```python
import os

kaggle_token_ready = bool(os.environ.get("KAGGLE_API_TOKEN"))
try:
    from google.colab import userdata
    token = userdata.get("KAGGLE_API_TOKEN")
    if token:
        os.environ["KAGGLE_API_TOKEN"] = token
        kaggle_token_ready = True
except Exception:
    # 非 Colab、未建立 Secret，或使用公開 Dataset 時都可略過。
    pass

print("Kaggle private/consent resource token:", "READY" if kaggle_token_ready else "NOT SET (public dataset mode can continue)")
```

## 4. 產生健檢結果


```python
checks = {
    "Python >= 3.10": sys.version_info >= (3, 10),
    "KaggleHub import": "kagglehub" in versions,
    "TensorFlow import": "tensorflow" in versions,
    "scikit-learn import": "sklearn" in versions,
    "Gradio import": "gradio" in versions,
    "GPU available (optional)": gpu_name != "CPU only",
}

for label, passed in checks.items():
    status = "PASS" if passed else ("OPTIONAL" if label.startswith("GPU") else "CHECK")
    print(f"{status:8} | {label}")

required_ok = all(value for key, value in checks.items() if not key.startswith("GPU"))
print("\nResult:", "READY FOR ROTATION" if required_ok else "ASK THE INSTRUCTOR FOR HELP")
```
