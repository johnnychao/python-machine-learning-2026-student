# Python Machine Learning 3rd ed. 環境檢查

先執行下方 cell，確認常用套件版本。Colab 上若套件缺漏，請依課堂指示安裝。


```python
import importlib
import platform
import sys

print("Python:", sys.version)
print("Platform:", platform.platform())

packages = [
    ("numpy", "numpy"),
    ("pandas", "pandas"),
    ("matplotlib", "matplotlib"),
    ("sklearn", "scikit-learn"),
    ("scipy", "scipy"),
    ("tensorflow", "tensorflow"),
    ("tensorflow_datasets", "tensorflow-datasets"),
    ("gym", "gym"),
]

for import_name, pip_name in packages:
    try:
        module = importlib.import_module(import_name)
        print(f"{pip_name}: {getattr(module, '__version__', 'installed')}")
    except Exception as exc:
        print(f"{pip_name}: not available ({type(exc).__name__})")
```
